Question 1: What is the difference between K-Means and Hierarchical Clustering?
Provide a use case for each

1. K-Means Clustering
Ans: Choose the number of clusters K.

Randomly initialize K centroids.

Assign each data point to the nearest centroid.

Recalculate the centroid of each cluster.

Repeat until centroids stop changing.

2. Hierarchical Clustering
Ans: Agglomerative (Bottom-Up) – Start with each point as a cluster and merge them.

Divisive (Top-Down) – Start with one cluster and split it.

Treat each data point as its own cluster.

Calculate distance between clusters.

Merge the two closest clusters.

Repeat until all points belong to one cluster.

Question 2: Explain the purpose of the Silhouette Score in evaluating clustering
algorithms.


Ans: The Silhouette Score is a metric used to evaluate how well a clustering algorithm has grouped the data. It measures how similar a data point is to its own cluster compared to other clusters

Purpose of Silhouette Score

The main purpose is to determine the quality of clustering by checking:

Cluster cohesion – how close data points are within the same cluster.

Cluster separation – how far clusters are from each other.

Question 3: What are the core parameters of DBSCAN, and how do they influence the
clustering process?

Ans: DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is a clustering algorithm that groups together points that are closely packed together and marks points in low-density areas as noise (outliers).

1. Epsilon (ε or eps)

Definition:
The maximum distance between two data points for them to be considered neighbors.

Influence on Clustering:

Small eps

Creates many small clusters

May mark many points as noise

Large eps

Clusters may merge together

Fewer clusters will be formed


2. MinPts (Minimum Points)

Definition:
The minimum number of points required within the eps radius to form a dense region (cluster).

Influence on Clustering:

Low MinPts

More clusters created

May include noise points in clusters

High MinPts

Clusters become stricter

More points labeled as noise

Question 4: Why is feature scaling important when applying clustering algorithms like K-Means and DBSCAN?

Ans: 1. Prevents Feature Dominance

If one feature has a much larger scale than others, it will dominate the distance calculation.

2. Improves Distance-Based Calculations

K-Means

DBSCAN

Hierarchical Clustering

use distance metrics (Euclidean, Manhattan, etc.).

3. Produces Better Cluster Structure

Without scaling:

Clusters may form incorrectly

Some variables may be ignored

With scaling:

Data points are distributed more uniformly

Clusters become more meaningful and accurate

Question 5: What is the Elbow Method in K-Means clustering and how does it help
determine the optimal number of clusters?

K-Means requires us to choose K (number of clusters) before training the model.
The Elbow Method helps find the best K value by analyzing how clustering performance changes as K increases.

WCSS (Within-Cluster Sum of Squares) measures the total distance between each point and its cluster centroid.

Lower WCSS = better clustering

As K increases, WCSS always decreases

Before the elbow → adding clusters significantly improves clustering.

After the elbow → improvement becomes minimal.

Question 6: Generate synthetic data using make_blobs(n_samples=300, centers=4),
apply KMeans clustering, and visualize the results with cluster centers.
(Include your Python code and output in the code box below.)

````python
# Import required libraries
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Generate synthetic dataset
X, y = make_blobs(n_samples=300, centers=4, random_state=42)

# Apply K-Means clustering
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(X)

# Predicted cluster labels
labels = kmeans.labels_

# Cluster centers
centers = kmeans.cluster_centers_

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=50)

# Plot cluster centers
plt.scatter(centers[:, 0], centers[:, 1], c='red', marker='X', s=200, label='Centroids')

plt.title("K-Means Clustering on make_blobs Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()

plt.show()
```
````

Question 7: Load the Wine dataset, apply StandardScaler , and then train a DBSCAN
model. Print the number of clusters found (excluding noise).
(Include your Python code and output in the code box below.)

````python
```python
# Import required libraries
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import numpy as np

# Load the Wine dataset
wine = load_wine()
X = wine.data

# Apply StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train DBSCAN model
dbscan = DBSCAN(eps=1.5, min_samples=5)
labels = dbscan.fit_predict(X_scaled)

# Count clusters (excluding noise label -1)
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

print("Number of clusters found (excluding noise):", n_clusters)
```

**Sample Output:**

```
Number of clusters found (excluding noise): 3
```
````

Question 8: Generate moon-shaped synthetic data using make_moons(n_samples=200, noise=0.1), apply DBSCAN, and highlight the outliers in the plot. (Include your Python code and output in the code box below.)

````python
```python
# Import required libraries
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import numpy as np

# Generate moon-shaped synthetic data
X, y = make_moons(n_samples=200, noise=0.1, random_state=42)

# Apply DBSCAN clustering
dbscan = DBSCAN(eps=0.2, min_samples=5)
labels = dbscan.fit_predict(X)

# Identify outliers (noise points labeled as -1)
outliers = labels == -1

# Plot clusters
plt.scatter(X[~outliers, 0], X[~outliers, 1], c=labels[~outliers], cmap='viridis', s=50)

# Highlight outliers
plt.scatter(X[outliers, 0], X[outliers, 1], c='red', marker='x', s=100, label='Outliers')

plt.title("DBSCAN Clustering on make_moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()

plt.show()
```

**Expected Output:**

```text
A scatter plot showing two moon-shaped clusters. Points belonging to clusters
are colored differently, while outliers (noise points) are marked with red 'x'.
```
````

Question 9: Load the Wine dataset, reduce it to 2D using PCA, then apply
Agglomerative Clustering and visualize the result in 2D with a scatter plot.
(Include your Python code and output in the code box below.)

````python
```python
# Import required libraries
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

# Load Wine dataset
wine = load_wine()
X = wine.data

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Reduce dimensions to 2 using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Apply Agglomerative Clustering
agg_cluster = AgglomerativeClustering(n_clusters=3)
labels = agg_cluster.fit_predict(X_pca)

# Visualize clusters
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', s=50)

plt.title("Agglomerative Clustering on Wine Dataset (PCA 2D)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```

**Sample Output:**

```
A 2D scatter plot showing the Wine dataset reduced to two principal components.
The points are colored into 3 clusters representing the groups identified by
Agglomerative Clustering.
```
````

1. Clustering Algorithm Choice

I would primarily use K-Means clustering because:

Additionally, I may test:

2. Data Preprocessing

Before clustering, the data must be cleaned and transformed.

Steps:

Handle Missing Values

Numeric features → fill with mean or median

Categorical features → fill with most frequent value

Feature Encoding

Convert categorical variables (gender, location, category preferences) using One-Hot Encoding.

Feature Scaling

Apply StandardScaler or MinMaxScaler because clustering uses distance-based calculations.

Feature Selection

Important features might include:

Purchase frequency

Average order value

3. Determining the Optimal Number of Clusters

To find the best K value, I would use:

Elbow Method

Plot K vs WCSS (Within Cluster Sum of Squares)

Choose the elbow point

Silhouette Score

Measure how well-separated clusters are

Higher score → better clustering



````python
```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Example synthetic customer dataset
X, _ = make_blobs(n_samples=300, centers=4, random_state=42)

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Determine clusters using K-Means
kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X_scaled)

# Plot clusters
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=labels, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            c='red', marker='X', s=200)

plt.title("Customer Segmentation using K-Means")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

**Sample Output**

```
Scatter plot showing customers grouped into 4 clusters with red 'X'
indicating cluster centers.
```
````
